### 5k AF2 HAMP models

```python
def parse_hamp_msa_fasta(path):
    records = {}
    cur_idx = None
    cur_def = None
    seq_parts = []

    with open(path, "r") as f:
        for line in f:
            line = line.rstrip("\n")
            if not line:
                continue

            if line.startswith(">"):
        # flush previous
                if cur_idx is not None:
                    records[cur_idx] = (cur_def, "".join(seq_parts))
                cur_def = line[1:]  # without ">"
                # index is the number before first "|"
                cur_idx = int(cur_def.split("|", 1)[0])
                seq_parts = []
            else:
                # keep gaps, remove whitespace just in case
                seq_parts.append(line.strip())

    # flush last
    if cur_idx is not None:
        records[cur_idx] = (cur_def, "".join(seq_parts))

    return records
```

```python
# measured 5k models
# generated with https://github.com/labstructbioinf/HAMPpred/blob/master/notebooks/steps/1_parse_AF2.ipynb
af2_5k_df = pd.read_pickle(os.path.join(data_dir, 'af2_clean_model_1.p'))
af2_5k_df['rot_mean'] = af2_5k_df['rot_mean'] / 2
af2_5k_df.head(3)
```

```python
# get MSA with domain annotations (DHp etc.)
seq_dict = parse_hamp_msa_fasta(os.path.join(data_dir, 'hamp_msa.fasta'))
```

```python
def domain_label(defline):
    if "DHpdom" in defline:
        return "DHpdom"
    if "MCP" in defline:
        return "MCP"
    return "unknown"

labels = [domain_label(seq_dict[i][0]) for i in af2_5k_df.index]

# add annotations
af2_5k_df['domain'] = labels
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import gaussian_kde

# columns: 'rot_mean' and 'domain'
rm = af2_5k_df["rot_mean"].dropna().to_numpy()
xmin, xmax = rm.min(), rm.max()
xs = np.linspace(xmin, xmax, 500)

plt.figure()

for dom in sorted(af2_5k_df["domain"].dropna().unique()):
    if dom == 'unknown': continue
    vals = af2_5k_df.loc[af2_5k_df["domain"] == dom, "rot_mean"].dropna().to_numpy()
    if vals.size < 2:
        continue
    kde = gaussian_kde(vals)
    plt.plot(xs, kde(xs), label="{} (n={})".format(dom, vals.size))

plt.xlabel("rot_mean")
plt.ylabel("density")
plt.legend()
plt.tight_layout()
plt.show()
```
